## Top 10 Manufacturers — BEV vs PHEV  

```python
top_makes = df['Make'].value_counts().head(10).index

ev_type_make_top10 = pd.crosstab(
    df['Make'],
    df['Electric Vehicle Type']
).loc[top_makes]

ev_type_make_top10
```

```python
ev_type_make_top10.plot(kind='bar', figsize=(12, 6))

plt.title('BEV vs PHEV Distribution by Top 10 Manufacturers')
plt.xlabel('Manufacturer')
plt.ylabel('Number of Vehicles')
plt.xticks(rotation=45)
plt.legend(title='Electric Vehicle Type')

plt.show()
```

#### Tesla has the highest number of BEVs among the top 10 manufacturers, with 118,148 vehicles, while Toyota and BMW have a higher number of PHEVs compared with their BEV counts. Overall, most of the top manufacturers have more BEVs than PHEVs.  

## EV Type Trend by Model Year  

```python
year_ev_type = pd.crosstab(
    df['Model Year'],
    df['Electric Vehicle Type']
)

year_ev_type
```

```python
year_ev_type.plot(
    kind='line',
    figsize=(12, 6),
    marker='o'
)

plt.title('BEV vs PHEV Trend by Model Year')
plt.xlabel('Model Year')
plt.ylabel('Number of Vehicles')
plt.legend(title='Electric Vehicle Type')

plt.show()
```

#### BEVs show a strong increase in recent model years, particularly from 2018 onward, with the highest count in 2023 at 53,152 vehicles. PHEVs also increased over time, reaching their highest count in 2024 at 10,448 vehicles, but remained considerably lower than BEVs in most recent model years.  

## Average Electric Range by EV Type  

```python
avg_range_by_type = range_df.groupby(
    'Electric Vehicle Type'
)['Electric Range'].mean()

avg_range_by_type
```

```python
plt.figure(figsize=(8, 6))

avg_range_by_type.plot(kind='bar')

plt.title('Average Electric Range by EV Type')
plt.xlabel('Electric Vehicle Type')
plt.ylabel('Average Electric Range (miles)')
plt.xticks(rotation=0)

plt.show()
```

#### Battery Electric Vehicles (BEVs) have a substantially higher average electric range of approximately 200 miles compared with PHEVs, which have an average electric range of approximately 32 miles. This highlights a significant difference in recorded electric range between the two vehicle types.  

## T-Test: BEV vs PHEV Electric Range 

```python
bev_range = range_df[
    range_df['Electric Vehicle Type'] == 'Battery Electric Vehicle (BEV)'
]['Electric Range']

phev_range = range_df[
    range_df['Electric Vehicle Type'] == 'Plug-in Hybrid Electric Vehicle (PHEV)'
]['Electric Range']
```

```python
t_stat, p_value = st.ttest_ind(
    bev_range,
    phev_range,
    equal_var=False
)

t_stat, p_value
```

#### The t-test results indicate a statistically significant difference in electric range between BEVs and PHEVs, with BEVs having a substantially higher average electric range than PHEVs.
